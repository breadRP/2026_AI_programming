## 실습 문제: 섭씨-화씨 양방향 온도 변환기 제작 및 테스트

**설명**: 이전 실습을 확장하여, 이번에는 섭씨를 화씨로, 그리고 화씨를 섭씨로 변환하는 두 개의 개별 함수를 만듭니다. 함수를 기능 단위로 분리하는 연습을 하고, 각 함수가 올바르게 작동하는지 확인하기 위한 간단한 테스트 코드를 작성하여 모듈화된 프로그래밍의 기초를 학습합니다.

$$F = C \times \frac{9}{5} + 32 \quad \text{그리고} \quad C = (F - 32) \times \frac{5}{9}$$

**요구사항**:

  - **함수 1**: `celsius_to_fahrenheit` 함수를 완성하여 섭씨 온도를 화씨 온도로 변환하세요.
  - **함수 2**: `fahrenheit_to_celsius` 함수를 완성하여 화씨 온도를 섭씨 온도로 변환하세요.
  - **테스트 코드**: 제공된 테스트 코드 섹션에서 각 함수를 호출하여 변환이 정확한지 확인하고 결과를 출력하세요. (예: 0°C → 32°F, 212°F → 100°C)



```python

def celsius_to_fahrenheit(celsius: float) -> float:
    """섭씨(Celsius)를 화씨(Fahrenheit)로 변환합니다."""
    # 이곳에 섭씨 -> 화씨 변환 코드를 작성하세요.
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

def fahrenheit_to_celsius(fahrenheit: float) -> float:
    """화씨(Fahrenheit)를 섭씨(Celsius)로 변환합니다."""
    # 이곳에 화씨 -> 섭씨 변환 코드를 작성하세요.
    celsius = (fahrenheit - 32) * 5/9
    return celsius
```

```python
# 테스트 1: 0°C가 32°F로 변환되는지 확인
c_temp = 0
f_expected = 32
f_actual = celsius_to_fahrenheit(c_temp)
print(f"{c_temp}°C -> {f_actual}°F (예상: {f_expected}°F)")
print(f"테스트 통과: {f_actual == f_expected}")

print("-" * 20)

# 테스트 2: 212°F가 100°C로 변환되는지 확인
f_temp = 212
c_expected = 100
c_actual = fahrenheit_to_celsius(f_temp)
print(f"{f_temp}°F -> {c_actual}°C (예상: {c_expected}°C)")
print(f"테스트 통과: {c_actual == c_expected}")
```

## 실습 문제: 부동소수점의 안전한 비교 함수 구현하기
**설명**: 컴퓨터는 십진수 소수를 이진수로 변환하는 과정에서 미세한 오차를 가집니다. 이로 인해 `0.1 + 0.2`는 정확히 `0.3`이 아니므로 `==` 연산자로 비교하면 `False`를 반환합니다. 이 문제를 해결하기 위해, 두 숫자의 차이가 아주 작은 값(허용 오차, 엡실론)보다 작은지를 확인하는 방법이 사용됩니다. 두 부동소수점 숫자를 안전하게 비교하는 함수를 직접 구현해 보세요.

$$ |a - b| < \epsilon $$

**요구사항**:
- `are_floats_equal`이라는 이름의 함수를 완성하여, 두 부동소수점 `a`와 `b`의 차이(절댓값)가 주어진 허용 오차 `epsilon`보다 작은지 검사하세요.
- `a`에 `0.1 + 0.2`를, `b`에 `0.3`을 할당하세요.
- 완성된 함수를 사용하여 `a`와 `b`가 실질적으로 같은지 확인하고 결과를 출력하세요.



```python
def are_floats_equal(a: float, b: float, epsilon: float = 1e-9) -> bool:
    """
    두 부동소수점 숫자가 허용 오차(epsilon) 내에서 같은지 확인합니다.

    Args:
        a (float): 첫 번째 부동소수점 숫자.
        b (float): 두 번째 부동소수점 숫자.
        epsilon (float): 허용되는 최대 오차. 기본값은 1e-9입니다.

    Returns:
        bool: 두 숫자가 허용 오차 내에서 같으면 True, 그렇지 않으면 False.
    """
    # 이 부분을 구현하세요.
    return abs(a - b) < epsilon
```

```python
# 테스트 코드
num1 = 0.1 + 0.2
num2 = 0.3

# 함수를 호출하여 결과를 확인하고 출력하세요.
result = are_floats_equal(num1, num2)
print(f"0.1 + 0.2와 0.3은 같은가? {result}")
```

## 실습 문제: 부동소수점 누적 오차 시뮬레이션
**설명**: 단일 부동소수점 연산의 오차는 매우 작지만, 이러한 작은 오차들이 반복문 안에서 수천, 수만 번 더해지면 무시할 수 없는 **누적 오차(Accumulated Error)** 로 커질 수 있습니다. 이번 실습에서는 `for` 루프를 사용해 작은 소수(`0.1`)를 여러 번 더하는 간단한 시뮬레이션을 통해, 이론적인 결과와 실제 계산 결과 사이에 오차가 어떻게 발생하는지 직접 확인해 보겠습니다. 💻

$$ \sum_{i=1}^{N} x \neq N \times x \quad (\text{in floating-point arithmetic}) $$

**요구사항**:
- `accumulate_error` 함수 내에서 `for` 루프를 사용해 `increment` 값을 `iterations` 횟수만큼 `total`에 더하는 코드를 완성하세요.
- `0.1`을 `10000`번 더한 결과를 함수를 호출하여 얻으세요.
- 수학적으로 예상되는 결과(`0.1 * 10000`)와 루프를 통해 얻은 실제 결과를 비교하고, 그 차이(누적 오차)를 출력하여 확인하세요.


```python

def accumulate_error(increment: float, iterations: int) -> float:
    """
    주어진 increment 값을 iterations 횟수만큼 반복해서 더하여
    부동소수점의 누적 오차를 확인합니다.

    Args:
        increment (float): 각 반복마다 더할 값.
        iterations (int): 반복 횟수.

    Returns:
        float: 누적된 합계.
    """
    total = 0.0
    # for 루프를 사용하여 increment를 total에 iterations 만큼 더하세요.
    # 이 부분을 구현하세요.
    for _ in range(iterations):
        total += increment
    return total

```

```python
# 시뮬레이션 파라미터
inc = 0.1
iters = 10000

# 함수 호출 및 결과 계산
actual_sum = accumulate_error(inc, iters)
expected_sum = inc * iters
error = actual_sum - expected_sum

print(f"루프를 이용한 계산 결과: {actual_sum}")
print(f"수학적 예상 결과: {expected_sum}")
print(f"누적 오차: {error}")
```
